```python
#print 1 to n odd numbers
n=int(input())
for i in range(1,n+1,2):
  print(i,end=" ")
```

```python
#print multiplication table of a given number
n=int(input())
for i in range(1,11,1):
  print(n,"*",i,"=",n*i)
```

```python
#check whether the number is prime or not
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("it is not prime")
    break
else:
    print("it is prime")
```

```python
#print all the prime numbers between given 2 numbers
m=int(input())
n=int(input())
for i in range(m,n+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end=" ")
```

```python
#print the sum of digits of a given number
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```
