```python
#finding the largest number
a=int(input("enter a value: "))
b=int(input("enter another value: "))
c=int(input("enter another value: "))
if(a>b and a>c):
  print(f"{a} is greater")
elif(b>a and b>c):
  print(f"{b}is greater")
else:
  print(f"{c} is greater")
```

```python
#print FIZZ if a number is divisible by 3,print BUZZ if a number is divisible by 5,print FIZZBUZZ if a number is divisible by both 3 & 5
a=int(input("enter a value: "))
if(a%3==0 and a%5!=0):
  print("FIZZ")
elif(a%5==0 and a%3!=0):
  print("BUZZ")
elif(a%3==0 and a%5==0):
  print("FIZZBUZZ")
```

```python
#day program on match case
day=int(input("enter a value: "))
match day:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("invalid choice")

```
